## Municipal Recycling

```python
recycling_rate["TIME_PERIOD"].unique()
```

```python
recycling_rate_2024 = recycling_rate[
    recycling_rate["TIME_PERIOD"] == 2024
].copy()

recycling_rate_2024.shape
```

```python
recycling_rate_eu27_2024 = recycling_rate_2024[
    recycling_rate_2024["geo"].isin(eu27)
].copy()

recycling_rate_eu27_2024.shape
```

```python
#handling missing data

missing_recycling_countries = set(eu27) - set(recycling_rate_eu27_2024["geo"])

missing_recycling_countries
```

```python
# Need to create a dataset that uses theblatest available value up to 2024 for every EU country.

recycling_rate_eu27_all = recycling_rate[
    recycling_rate["geo"].isin(eu27) & (recycling_rate["TIME_PERIOD"] <= 2024)
].copy()

#Newest year first

recycling_rate_eu27_all = recycling_rate_eu27_all.sort_values(["geo", "TIME_PERIOD"], ascending=[True, False]).reset_index(drop=True)

recycling_rate_latest = recycling_rate_eu27_all.drop_duplicates(subset=["geo"], keep="first").reset_index(drop=True)

recycling_rate_latest.shape
```

```python
recycling_rate_latest[
    ["geo",
     "Geopolitical entity (reporting)",
     "TIME_PERIOD",
     "OBS_VALUE"]
].sort_values("TIME_PERIOD")
```

```python
recycling_rate_clean = recycling_rate_latest[[
    "geo",
    "Geopolitical entity (reporting)",
    "TIME_PERIOD",
    "OBS_VALUE"
]].copy()

recycling_rate_clean = recycling_rate_clean.rename(columns={
     "geo": "country_code",
     "Geopolitical entity (reporting)": "country_name",
     "TIME_PERIOD": "year",
     "OBS_VALUE": "municipal_recycling_rate"
})
```

```python
recycling_rate_clean = (recycling_rate_clean.sort_values("country_name").reset_index(drop=True))

recycling_rate_clean.isnull().sum()
```

```python
eqli_data = pd.merge(
    eqli_data,
    recycling_rate_clean[["country_code", "municipal_recycling_rate"]],
    on="country_code",
    how="inner"
)

eqli_data.shape
```

```python
recycling_rate_clean.to_csv("../data/processed/recycling_rate_latest.csv", index=False)

eqli_data.to_csv(
    "../data/processed/eqli_master.csv",
    index=False
)
```
