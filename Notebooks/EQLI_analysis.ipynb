# GDP per Capita

```python
gdp = pd.read_csv("../data/raw/gdp_per_capita_pps.csv")

gdp.head()
```

```python
gdp_2024 = gdp[gdp["TIME_PERIOD"] == 2024].copy()

gdp_2024.head()
```

```python
gdp_2024[[
    "geo",
    "Geopolitical entity (reporting)",
    "TIME_PERIOD",
    "OBS_VALUE"
]].head(15)
```

```python
eu27 = [
    "AT", "BE", "BG", "HR", "CY", "CZ", "DK",
    "EE", "FI", "FR", "DE", "EL", "HU", "IE",
    "IT", "LV", "LT", "LU", "MT", "NL", "PL",
    "PT", "RO", "SK", "SI", "ES", "SE"
]
```

```python
gdp_eu27 = gdp_2024[gdp_2024["geo"].isin(eu27)].copy()

gdp_eu27[[
    "geo",
    "Geopolitical entity (reporting)",
    "TIME_PERIOD",
    "OBS_VALUE"
]].head()
```

```python
gdp_eu27.shape
```

```python
gdp_eu27["geo"].nunique()
```

```python
gdp.columns
```

```python
gdp_clean = gdp_eu27[
    ["geo", "Geopolitical entity (reporting)", "TIME_PERIOD", "OBS_VALUE"]
].copy()

gdp_clean.head()
```

```python
gdp_clean = gdp_clean.rename(columns={
    "geo": "country_code",
    "Geopolitical entity (reporting)": "country_name",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "gdp_per_capita_pps"
})

gdp_clean.head()
```

```python
gdp_clean.to_csv("../data/processed/gdp_clean.csv", index=False)
```

# Employment Rate

```python
employment = pd.read_csv("../data/raw/employment_rate.csv")

employment.head()
```

```python
employment.shape
employment.columns
employment.info()
```

```python
employment_2024 = employment[employment["TIME_PERIOD"] == 2024].copy()
employment_2024.shape
```

```python
employment_eu27 = employment_2024[employment_2024["geo"].isin(eu27)].copy()

employment_eu27.shape
```

```python
employment_clean = employment_eu27[
    [
        "geo", "Geopolitical entity (reporting)", "TIME_PERIOD", "OBS_VALUE"
    ]
].copy()
```

```python
employment_clean = employment_clean.rename(columns={
    "geo": "country_code",
    "Geopolitical entity (reporting)": "country_name",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "employment_rate"
})
```

```python
employment_clean = (
    employment_clean.sort_values("country_name").reset_index(drop=True))

employment_clean.head()
```

```python
employment_clean.isnull().sum()
```

```python
employment_clean.to_csv("../data/processed/employment_clean.csv", index=False)
```

## Merging GDP and Employment


```python
eqli_data = pd.merge(
    gdp_clean[["country_code", "country_name",  "gdp_per_capita_pps"]],
    employment_clean[["country_code", "employment_rate"]],

    on="country_code",
    how="inner"

)

eqli_data.head()
```

```python
eqli_data.shape
eqli_data.isnull().sum()
```

```python
eqli_data[["country_name", "gdp_per_capita_pps", "employment_rate"]].head(10)
```

```python
eqli_data["country_code"].nunique()
```

```python
set(eu27) - set(eqli_data["country_code"])
```

```python
eqli_data.to_csv(
    "../data/processed/eqli_master.csv",
    index=False
)
```
